# Exercício 17: Conte a frequência de cada palavra em uma string fornecida pelo usuário.

```python
frase = str(input("Digite uma frase: "))

palavras = frase.split()
contagem = {}

for palavra in palavras:

  if palavra in contagem:
    contagem[palavra] +=1

  else:
    contagem [palavra] = 1

print (contagem)
```

Explicação: Comecei o programa criando a variável frase, do tipo str (string) que recebe a função input para permitir que o usuário digite uma frase. Em seguida, criei a variável palavras, que recebe a função split que é utilizada na variável frase. Essa função serve para separar a frase em palavras e transformar o texto em uma lista. Depois, criei a variável contagem, que recebe um dicionário vazio {}, onde serão armazenadas as palavras e suas respectivas quantidades. Em seguida, utilizei a estrutura for para percorrer cada palavra dentro da variável palavras. Dentro do for, utilizei a estrutura de repetição if para verificar se a palavra já existe dentro do dicionário contagem. Se a palavra já existir, utilizo contagem[palavra] += 1 para adicionar 1 a quantidade de vezes que essa palavra apareceu. Caso a palavra ainda não exista no dicionário, utilizo o else e contagem[palavra] = 1 para adicionar a palavra ao dicionário e definir sua quantidade como 1. Por fim, utilizei a função print para exibir o dicionário contagem, mostrando cada palavra e o número de vezes que ela apareceu na frase.

Dentro do for, utilizei a estrutura if para verificar se o número num é divisível pelo número i. Para isso, utilizei o operador %, que verifica o resto da divisão. Se num % i == 0, significa que a divisão é exata, então utilizo contagem += 1 para adicionar 1 à quantidade de divisões. Em seguida, utilizei a função print junto com o f-string para exibir o número digitado e a quantidade de vezes que ele foi divisível. Depois, utilizei novamente a estrutura if para verificar se a variável contagem é igual a 2. Se for igual a 2, significa que o número possui somente dois divisores, 1 e ele mesmo, então ele é um número primo. Caso contrário, utilizo o else para informar que o número não é primo.

# Exercício 21: Implemente a busca binária para encontrar um elemento em uma lista ordenada.

```python
lista = [1, 3, 5, 7, 9, 11, 13]

numero = int(input("Digite o número que deseja encontrar: "))

inicio = 0
fim = len(lista) - 1

while inicio <= fim:
  meio = (inicio + fim) // 2


  if lista[meio] == numero:
    print("Elemento encontrado na posição", meio)
    break

  elif lista[meio] < numero:
    inicio = meio + 1

  else:
    fim = meio - 1

else:
  print("Elemento não encontrado")
```

Explicação: Comecei o programa criando a variável lista, que recebe uma lista de números inteiros ordenados em ordem crescente. Em seguida, criei a variável numero, do tipo int, que recebe a função input para permitir que o usuário digite o número que deseja encontrar na lista. Depois, criei as variáveis inicio e fim. A variável inicio recebe 0, pois representa a primeira posição da lista. Já a variável fim recebe len(lista) - 1, utilizando a função len para descobrir a quantidade de elementos da lista e subtraindo 1, pois as posições da lista começam no 0. Em seguida, utilizei a estrutura while inicio <= fim para repetir a busca enquanto ainda existir uma parte da lista para ser analisada. Dentro do while, criei a variável meio, que recebe (inicio + fim) // 2. O operador // realiza uma divisão inteira, permitindo encontrar a posição que está no meio da parte da lista que está sendo analisada. Depois, utilizei a estrutura if para verificar se o elemento que está na posição meio é igual ao número que o usuário digitou. Para isso, utilizo lista[meio] == numero. Se forem iguais, utilizo a função print() para informar que o elemento foi encontrado e mostrar sua posição. Em seguida, utilizo o break para interromper o while, pois o número já foi encontrado. Caso o número da posição meio seja menor que o número procurado, utilizo o elif. Nesse caso, faço inicio = meio + 1, descartando a parte esquerda da lista e continuando a busca somente na parte direita. Isso é possível porque a lista está ordenada. Caso o número da posição meio seja maior que o número procurado, utilizo o else e faço fim = meio - 1. Dessa forma, descarto a parte direita da lista e continuo procurando somente na parte esquerda. Por fim, o else que está alinhado com o while é executado quando o while termina sem encontrar o número. Nesse caso, utilizo a função print para informar que o elemento não foi encontrado.

# Exercicio 22: Crie uma função para criptografar uma mensagem usando a cifra de César.

```python
def criptografar (texto, deslocamento):
  resultado = ''

  for letra in texto:
    if letra.isalpha():
      nova_letra = chr((ord(letra) - ord("a") + deslocamento) % 26 + ord("a"))
      resultado += nova_letra

    else:
      resultado += letra

  return resultado

texto = str(input("Digite a mensagem: "))
deslocamento = int(input("Digite o deslocamento: "))

print(f"Mensagem criptografada: {criptografar(texto, deslocamento)}")

```

Explicação: Comecei o programa criando a função criptografar, que recebe dois valores: texto, que representa a mensagem que será criptografada, e deslocamento, que indica quantas posições cada letra irá avançar no alfabeto. Em seguida, criei a variável resultado, que começa como uma string vazia e será utilizada para guardar a mensagem já criptografada. Depois, utilizei o for para percorrer cada letra que existe dentro do texto. Dentro do for, utilizei o if junto com a função isalpha para verificar se o caractere é uma letra. Se for uma letra, criei a variável nova_letra, que realiza a criptografia utilizando as funções ord e chr. A função ord transforma a letra em um número correspondente na tabela de caracteres, e a função chr faz o processo contrário, transformando o número novamente em uma letra. O % 26 serve para fazer com que, quando a letra passar do z, ela volte para o início do alfabeto. Depois, utilizo resultado += nova_letra para adicionar a nova letra criptografada ao resultado. Caso o caractere não seja uma letra, o else é executado e o caractere original é adicionado ao resultado. Dessa forma, espaços, números e outros caracteres são mantidos na mensagem. Por fim, utilizo return resultado para devolver a mensagem já criptografada pela função. Fora da função, criei a variável texto, do tipo str(string) que recebe a mensagem digitada pelo usuário, e a variável deslocamento, do tipo int, que recebe o número de posições que as letras devem avançar. Por último, utilizo a função print junto com o f-string para mostrar a mensagem criptografada.

# Exercício 23: Crie uma função para descriptografar uma mensagem cifrada com a cifra de César.

```python
def descriptografar (texto, deslocamento):
  resultado = ''

  for letra in texto:
    if letra.isalpha():
      nova_letra = chr((ord(letra) - ord("a") - deslocamento) % 26 + ord("a"))
      resultado += nova_letra

    else:
      resultado += letra

  return resultado

texto = str(input("Digite a mensagem criptografada: "))
deslocamento = int(input("Digite o deslocamento: "))

print(f"Mensagem descriptografada: {descriptografar(texto, deslocamento)}")
```

Explicação: Comecei o programa criando a função descriptografar, que recebe dois valores: texto, que representa a mensagem que foi criptografada, e deslocamento, que indica quantas posições cada letra deve voltar no alfabeto. Em seguida, criei a variável resultado, que começa como uma string vazia e será utilizada para guardar a mensagem já descriptografada. Depois, utilizei o for para percorrer cada letra que existe dentro do texto. Dentro do for, utilizei o if junto com a função isalpha para verificar se o caractere é uma letra. Se for uma letra, criei a variável nova_letra, que realiza a descriptografia utilizando as funções ord e chr. A diferença em relação à criptografia é que, nesse caso, utilizei - deslocamento, fazendo com que a letra volte algumas posições no alfabeto. A função ord transforma a letra em um número, enquanto a função chr transforma esse número novamente em uma letra. O % 26 serve para manter o resultado dentro das 26 letras do alfabeto. Por exemplo, se precisarmos voltar algumas posições a partir da letra a, o programa consegue continuar a partir do final do alfabeto. Depois, utilizo resultado += nova_letra para adicionar a letra descriptografada ao resultado. Caso o caractere não seja uma letra, o else é executado e o caractere original é adicionado ao resultado. Dessa forma, espaços, números e outros caracteres continuam iguais. Por fim, utilizo return resultado para devolver a mensagem já descriptografada pela função. Fora da função, criei a variável texto, que recebe a mensagem criptografada digitada pelo usuário, e a variável deslocamento, que recebe o número de posições que as letras devem voltar. Por último, utilizo a função print junto com o f-string para mostrar a mensagem descriptografada.

# Exercício 24: Implemente uma função para validar um CPF brasileiro.

```python
def validar_cpf (cpf):
  cpf = cpf.replace(".", "").replace("-", "")

  if len(cpf) != 11 or not cpf.isdigit():
    return False

  soma = 0
  for i in range(9):
    soma += int(cpf[i]) * (10 - i)

  resto = soma % 11

  if resto < 2:
    digito1 = 0
  else:
    digito1 = 11 - resto

  if digito1 != int(cpf[9]):
    return False

  soma = 0
  for i in range (10):
    soma += int(cpf[i]) * (11 - i)

  resto = soma % 11

  if resto < 2:
    digito2 = 0
  else:
    digito2 = 11 - resto

  if digito2 != int(cpf[10]):
    return False

  return True

cpf = (input("Digite seu CPF: "))

if validar_cpf (cpf):
  print("CPF válido")
else:
  print("CPF inválido")
```

Explicação: Comecei o programa criando a função validar_cpf, que recebe o CPF digitado pelo usuário. Caso alguma dessas condições não seja verdadeira, utilizo return False para informar que o CPF é inválido. Depois, criei a variável soma, do tipo int e que começa com valor 0, e uma estrutura for para calcular a soma necessária para verificar o primeiro dígito do CPF. Dentro do for, utilizo (cpf[i]) para transformar cada caractere do CPF em número e multiplico esse número pelo peso correspondente, que é calculado por (10 - i). O resultado de cada multiplicação é adicionado a variável soma. Em seguida, criei a variável resto, que recebe o resto da divisão de soma por 11, utilizando o operador %. Esse resto é utilizado para calcular o primeiro dígito verificador. Se o resto for menor que 2, o digito1 recebe 0. Caso contrário, digito1 recebe 11 - resto. Depois, comparo o digito1 calculado com o décimo número do CPF, que está na posição 9. Se os dois forem diferentes, utilizo return False, indicando que o CPF é inválido. Depois, faço praticamente o mesmo processo para verificar o segundo dígito. Primeiro, reinicio a variável soma com 0 e utilizo outro for, dessa vez percorrendo os 10 primeiros números do CPF. # Exercício 31: Implemente o algoritmo de ordenação por bolha (Bubble Sort).

```python
numeros = [5, 2, 8, 1, 3, 0, 11]

for i in range(len(numeros)):
    for a in range(len(numeros) - 1):
        if numeros[a] > numeros[a + 1]:
            numeros[a], numeros[a + 1] = numeros[a + 1], numeros[a]

print(numeros)
```

Explicação: Comecei o programa criando a variável numeros, que recebe uma lista com vários números inteiros que estão desorganizados. Em seguida, utilizei a estrutura for junto com range(len(numeros)) para repetir o processo de comparação várias vezes. A função len é utilizada para descobrir quantos elementos existem na lista. Depois, criei outro for dentro do primeiro for, utilizando range(len(numeros) - 1). Esse segundo for serve para percorrer a lista e comparar os números que estão lado a lado. O -1 é utilizado porque, dentro da comparação, será analisado o elemento atual junto com o próximo elemento. Dentro do segundo for, utilizei o if para verificar se o número que está na posição a é maior que o número que está na posição seguinte, representada por a + 1. Se numeros[a] > numeros[a + 1], significa que os dois números estão na ordem errada, já que tem que organizar a lista do menor para o maior. Quando isso acontece, utilizo numeros[a], numeros[a + 1] = numeros[a + 1], numeros[a] para trocar os dois números de posição. Dessa forma, o número menor passa para a esquerda e o maior passa para a direita. O primeiro for faz com que esse processo de comparação e troca seja repetido várias vezes. Isso é necessário porque uma única passagem pela lista não é suficiente para colocar todos os números na posição correta. Esse método é conhecido como Bubble Sort, ou ordenação por bolha. Por fim, utilizo a função print para mostrar a lista depois de todas as comparações e trocas.

# Exercício 46: Calcule a média e o desvio padrão de uma lista de números.

```python
numeros = [10, 20, 30, 40, 50]

media = sum(numeros) / len(numeros)

soma = 0

for numero in numeros:
    soma += (numero - media) ** 2

desvio = (soma / len(numeros)) ** 0.5

print(f"Média: {media}")
print(f"Desvio padrão:{desvio :.2f}")
```

Explicação: Comecei o programa criando a variável numeros, que recebe uma lista com vários números. Em seguida, utilizei a estrutura for para percorrer cada numero que existe dentro da lista. Dentro do for, utilizei soma += (numero - media) ** 2 para calcular a diferença entre cada número e a média, elevar essa diferença ao quadrado e adicionar o resultado à variável soma. Depois, criei a variável desvio, que recebe a divisão da soma pela quantidade de números da lista, utilizando len(numeros). Em seguida, utilizo ** 0.5 para calcular a raiz quadrada desse resultado, obtendo assim o desvio padrão. Por fim, utilizei a função print junto com o f-string para exibir a média e o desvio padrão. No desvio padrão, utilizei :.2f para mostrar somente dois números após o ponto.

# Exercício 48: Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
numeros = [5, 2, 4, 1, 3]

for i in range(1, len(numeros)):
    j = i

    while j > 0 and numeros[j] < numeros[j - 1]:
        numeros[j], numeros[j - 1] = numeros[j - 1], numeros[j]
        j -= 1

print(numeros)
```

Explicação: Comecei o programa criando a variável numeros, que recebe uma lista com números desorganizados. Em seguida, utilizei a estrutura for junto com range(1, len(numeros)) para percorrer a lista começando pela segunda posição. A função len serve para descobrir quantos elementos existem na lista. Dentro do for, criei a variável j, que recebe o valor de i. Essa variável será utilizada para controlar a posição do número que está sendo comparado e, se necessário, movê-lo para a posição correta. Depois, utilizei a estrutura while para verificar se j é maior que 0 e se o número que está na posição j é menor que o número que está na posição anterior. Se as duas condições forem verdadeiras, significa que os números estão na ordem errada. Dentro do while, utilizei numeros[j], numeros[j - 1] = numeros[j - 1], numeros[j] para trocar os dois números de posição. Dessa forma, o número menor vai sendo levado para a esquerda até encontrar a posição correta. Em seguida, utilizo j -= 1 para diminuir a posição de j e continuar comparando o número com os elementos anteriores. Por fim, utilizei a função print para exibir a lista depois que todos os números foram organizados.
