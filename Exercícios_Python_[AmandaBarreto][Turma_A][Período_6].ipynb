# Exercicio 3: Solicite o raio de um círculo e calcule sua área. Use a fórmula área = π * r².

```python
import math
raio = float(input("Digite o valor do raio: "))
area = math.pi * (raio** 2)
print(f"O valor da área é igual a {area: .2f}")
```

Logo após eu criei o cálculo, conforme a fórmula dada no exercício, utilizando a funcionalidade math.pi da biblioteca math, para o valor de pi multiplicando pelo raio ao quadrado. # Converta uma temperatura fornecida em Celsius para Fahrenheit usando a fórmula F = C * 9/5 + 32.

```python
celsius = float(input("Digite a temperatura em Celsius: "))
fahrenheit = celsius * (9/5) + 32
print(f"A temperatura em Fahrenheit é igual a {fahrenheit: .2f} ºF")
```

Eu comecei o exercício com a cria
ção de variavel celsius do tipo float, utilizei também a função input para que o usuário consiga digitar o valor que ele deseja em ºC para fazer a conversão. Logo após apliquei a fórmula de conversão de ºC para ºF, armazenando esse valor em uma variavel fahrenheit. Por último, utilizei a função print para exibir para o usuário o valor em Fahrenheit, utilizei tmabém a função f-string para formatar o valor em ºF para somente dois números após a vírgula
