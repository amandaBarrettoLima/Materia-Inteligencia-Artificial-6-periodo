### 4. Gere um array de 100 números inteiros aleatórios e encontre a mediana e a moda dos valores.

```python
import numpy as np

array = np.random.randint(1, 101, 100)
print(f"Array: {array}")

mediana = np.median(array)
valores, contagens = np.unique(array, return_counts=True)
moda = valores[np.argmax(contagens)]

print(f"Mediana: {mediana}")
print(f"Moda: {moda}")
```

Depois, utilizei a função print para exibir o array gerado. Em seguida, criei uma variável chamada mediana e utilizei a função np.median(array) para calcular a mediana dos números do array. A mediana representa o valor que fica no meio dos dados quando eles estão organizados em ordem. Depois, utilizei a função np.unique(array, return_counts=True) para encontrar os valores diferentes presentes no array e também contar quantas vezes cada um deles aparece. Os valores diferentes foram armazenados na variável valores e as quantidades de vezes que cada valor aparece foram armazenadas na variável contagens. Em seguida, criei a variável moda para descobrir qual número aparece mais vezes. Para isso, utilizei a função np.argmax(contagens), que encontra a posição onde está a maior quantidade de ocorrências. Depois, utilizei essa posição para acessar o valor correspondente dentro da variável valores. Por fim, utilizei a função print para exibir a mediana e a moda calculadas pelo programa.

### 7. Crie dois arrays de números aleatórios e encontre a correlação entre eles.

```python
import numpy as np

a = np.random.rand(15)
b = np.random.rand(15)

print(f"Array a: {a}")
print(f"\nArray b: {b}")

correlacao = np.corrcoef(a, b)[0, 1]

print(f"\nCorrelação: {correlacao}")
```

Como essa função retorna uma matriz 2x2 com as correlações entre os próprios arrays e entre eles, utilizei [0, 1] para acessar especificamente o valor que representa a correlação entre a e b. A correlação indica o quanto os dois conjuntos de dados possuem uma relação entre si, podendo variar de -1 até 1: valores próximos de 1 indicam uma relação positiva, valores próximos de -1 indicam uma relação negativa e valores próximos de 0 indicam pouca ou nenhuma relação linear. ### 9. Crie um array com 10 números aleatórios e normalize seus valores entre 0 e 1.

```python
import numpy as np

array = np.random.rand(10) * 100
print(f"Array: {array}")

array_normalizado = ((array - array.min()) / (array.max() - array.min()))
print(f"\nArray normalizado: {array_normalizado}")
```

Explicação: Comecei o programa importando a biblioteca NumPy e utilizei o as np para criar um apelido para facilitar sua utilização no código. Em seguida, criei uma variável chamada array e utilizei a função np.random.rand(10) para gerar um array com 10 números decimais aleatórios entre 0 e 1. Em seguida, criei uma variável chamada array_normalizado para armazenar os valores depois da normalização. Para fazer isso, utilizei a fórmula de normalização Min-Max, que serve para transformar os valores para uma escala entre 0 e 1. Primeiro, utilizei array.min() para encontrar o menor valor do array e subtraí esse valor de cada elemento com array - array.min(). Depois, utilizei array.max() para encontrar o maior valor e array.min() novamente para encontrar o menor, calculando a diferença entre eles. Por fim, dividi cada valor pela diferença entre o maior e o menor valor, fazendo com que o menor valor do array se torne 0 e o maior valor se torne 1, enquanto os outros valores ficam proporcionalmente entre eles. Por fim, utilizei a função print para exibir o array depois de normalizado.

### 11. Gere um array de 100 números inteiros e encontre o valor que mais se repete no array.

```python
import numpy as np

array = np.random.randint(1, 20, 100)
print(f"Array: \n{array}")

frequencia = np.bincount(array)
mais_frequente = np.argmax(frequencia)
print(f"Valor mais frequente: {mais_frequente}")
```

Essa função cria um novo array em que o índice representa o número e o valor naquela posição representa a quantidade de vezes que esse número apareceu. ### 14. Calcule a diferença entre dois arrays de números aleatórios de mesmo comprimento e encontre a média das diferenças.

Para isso, considero y[0] como sendo y e y[1] como sendo y'. A equação original pode ser reorganizada para y'' = -y, então criei a função equacao(x, y) que retorna [y[1], -y[0]]. O primeiro valor representa y' e o segundo representa y''. Em seguida, utilizei solve_ivp(equacao, [0, 10], [1, 0]) para resolver a equação no intervalo de x = 0 até x = 10. Em seguida, criei a variável resultados para armazenar os valores que estão depois do sinal de igualdade, ou seja, 7, 8 e 9. Depois, utilizei a função linalg.solve(matriz, resultados) para resolver o sistema de equações. Essa função recebe a matriz com os coeficientes e o array com os resultados e retorna os valores de x, y e z. O resultado foi armazenado na variável solucao. ### 8. Plote um gráfico de linha com intervalos de confiança (sombreamento) para a função y = sin(x).

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)

y = np.sin(x)

erro = 0.2

limite_inferior = y - erro
limite_superior = y + erro

plt.plot(x, y, label="y = sin(x)")

plt.fill_between(x, limite_inferior, limite_superior, alpha=0.3)

plt.title("Função y = sin(x) com intervalo de confiança")
plt.xlabel("x")
plt.ylabel("y")

plt.legend()

plt.show()
```

Explicação: Comecei o programa importando a biblioteca NumPy para trabalhar com os valores numéricos e utilizei o as np para facilitar sua utilização. Em seguida, importei o pyplot da biblioteca Matplotlib utilizando o as plt, que será utilizado para criar o gráfico. Em seguida, criei a variável y utilizando np.sin(x), calculando o seno de cada valor de x. Depois, defini um erro de 0.2 e criei dois limites: limite_inferior, que representa y - 0.2, e limite_superior, que representa y + 0.2. Esses dois limites formam a região do intervalo de confiança ao redor da função. Em seguida, utilizei plt.plot(x, y) para desenhar a linha da função. Depois, utilizei plt.fill_between() para preencher com sombreamento a região entre o limite inferior e o limite superior. Por fim, adicionei título, nomes aos eixos e uma legenda, utilizando plt.title(), plt.xlabel(), plt.ylabel() e plt.legend(). O plt.show() exibe o gráfico.

### 18. Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.

```python
import matplotlib.pyplot as plt

categorias = ["A", "B", "C", "D", "E"]
valores = [20, -15, 30, -10, 25]

cores = []

for valor in valores:
    if valor >= 0:
        cores.append("green")
    else:
        cores.append("red")

plt.bar(categorias, valores, color=cores)

plt.title("Valores positivos e negativos")
plt.xlabel("Categorias")
plt.ylabel("Valores")

plt.axhline(0)

plt.show()
```

Explicação: Comecei o programa importando o pyplot da biblioteca Matplotlib e utilizei o as plt para facilitar sua utilização. Em seguida, criei uma lista chamada categorias com cinco categorias e uma lista chamada valores contendo números positivos e negativos. Depois, criei uma lista vazia chamada cores. Utilizei um for para percorrer cada valor da lista valores. Dentro do for, utilizei uma estrutura if para verificar se o valor era maior ou igual a zero. Se fosse positivo, adicionamos "green" à lista de cores; caso contrário, adicionamos "red". Dessa forma, cada valor recebe uma cor de acordo com seu tipo. Depois, utilizei plt.bar() para criar o gráfico de barras e passei color=cores para utilizar as cores definidas anteriormente. Por fim, utilizei plt.axhline(0) para criar uma linha horizontal no valor zero, facilitando a visualização da separação entre valores positivos e negativos.

### 20. Adicione uma grade com espaçamento personalizado a um gráfico de linha para y = x2.

Depois, plt.grid() cria a grade baseada nessas marcações. Por fim, plt.show() exibe o gráfico.
